# Computational Analysis of Chloride in Meat Using the Mohr Method

## Project Overview

This project develops a Python-based computational model for the quantitative
analysis of chloride ions in meat using the Mohr argentometric titration method.

The project is based on a chloride-analysis experiment I performed during
high school. Because laboratory equipment is not currently available to me,
I recreated the calculation and data-analysis component computationally.

## Objectives

- Calculate chloride content from AgNO3 titration data
- Convert experimental measurements into quantitative results
- Analyze multiple samples computationally
- Visualize the relationship between titrant volume and chloride content
- Investigate the sensitivity of the calculated result to experimental inputs
- Explore measurement variability using basic statistical analysis

## Chemical Principle

Ag+ + Cl- → AgCl(s)

At the equivalence point, the amount of Ag+ reacting with chloride is
related to the amount of chloride present through a 1:1 stoichiometric ratio.

## Important Note

The numerical datasets used in the computational analysis are simulated
examples based on the experimental method. They are not presented as
measurements from the original laboratory experiment.

## Methodology

The computational workflow follows these steps:

1. Enter the AgNO3 concentration.
2. Enter the volume of AgNO3 used during titration.
3. Convert the titrant volume from mL to L.
4. Calculate the moles of AgNO3.
5. Use the 1:1 stoichiometric relationship to determine moles of Cl-.
6. Convert moles of Cl- to mass using its molar mass.
7. Calculate the percentage of chloride in the meat sample.
8. Repeat the calculation for multiple simulated samples.
9. Calculate the average and standard deviation.
10. Visualize the results using scientific graphs.
11. Perform sensitivity analysis on selected input variables.
